```python
from Bio.Seq import Seq
from Bio.SeqUtils import gc_fraction,molecular_weight
from Bio.SeqUtils import MeltingTemp as mt
print("DNA analysis starts soon(Initializing......)")
def analysis(user_input):
  dna=Seq(user_input)
  dna_template=dna.complement()
  mRNA=dna.transcribe() # Fixed: transcribe should be called without arguments
  polypeptide_chain=dna.translate(to_stop=True) #AFTER TRANSLATION  IN CYTOPLASM

  #GC PERCENTAGE CALCULATION
  gc_percent=gc_fraction(dna)*100
  #Tendem repeats check section
  is_slippage= False
  reason=""
  if ("CAG" * 31) in str(dna):
    is_slippage=True
    reason="CAG repeats causes DNA polymerase slippage that led to Huntington's risk"
  elif("CTG"*50) in str(dna):
    is_slippage=True
    reason="CTG repeats causes Myotonic Dystrophy risk"
  elif ("CGG"*200) in str(dna):
    is_slippage=True
    reason="CGG causes the Fragile X risk"
  if is_slippage:
    print(f"Results:DNA POLYMERASE occur because of {reason}")
  else:
    print("Results:Normal replication will occur")
 #THE MOLECULAR WEIGHT OF PROTEIN CALCULTAION
  mole_wght = 0.0 # Initialize mole_wght
  try:
    protein=dna.translate(to_stop=True)
    if len(protein)==0:
      mole_wght = 0.0 # Assign to mole_wght, not the function
    else:
      mole_wght=molecular_weight(protein,seq_type="protein")
  except Exception:
    mole_wght = 0.0 # Assign to mole_wght in case of exception
#MELTING TEMPETARATURE
  melting_temperature=mt.Tm_GC(dna)
#ECOLOGICAL NICHE DEPENDING ON MELTING TEMPERATURE AND GC PERCENNTAGE  BASED ON EVOLUTIONARY ND MOLECULAR BIOLOGY
#Evolutionary fact:High GC + HIGH Tm >>>>>>>>>>>>>>> Extreme heat Adaptation
  if gc_percent > 60 and melting_temperature >75:
    niche=("HYPERTHERMOPHILLIC / THERMOPHILIC NICHE \n"
            "[Fact]:Found in Hdrothermal Vents or Hot Springs e.g Thermus aquaticus"
            "[Evolutionary Adaptation]:High GC tripple Hydrogen Bonds and Elevtaed Tm stop DNA from Melting in Extreme Heat")
  elif 40 <= gc_percent <=60 and 55 <= melting_temperature <=75:
    niche=("MESOPHILIC NICHE\n"
            "[Fact]:Found in moderate in environment such as soils or mammalian hosts such as human colon ,consist of most of parasites such as E.coli"
            "[Evolutionary Adaptation]:Balanced GC-AT ratio optimized for stable replicationat standard Temperature(30℃-45℃)")
  else:
    niche=("UNUSUAL/HALOPHILIC OR ADAPTIVE NICHE/n"
          "[Fact]:Pontelial organism from high salt(Halophile) or deep-sea environment\n"
          "[Evolutionary Adaptation]:Discrerpancy btn GC percentage and Tm often hints a specific structural proteins protecting the Genome")
  return{"dna":dna,"template":dna_template,"mRNA":mRNA,"polypeptide_chain":polypeptide_chain,
           "gc":gc_percent,"molw":mole_wght,"tm":melting_temperature,"niche":niche}


#STARTING OF OF ANALYSIS
print("WELCOME IN DNA ANALYSER Version 2.003!")
print("Type EXIT/exit whenever you want to close the programme " )

while True:
    user_seq_input=input("Enter valid DNA sequence:  ").strip().upper()
    if user_seq_input.lower()=="exit": # Fixed: .lower() should be called as a method
      print("\n Thanks for using our analyser.Programme is closed!👏")
      break

    if not user_seq_input:
      print("You're not allowed to leave the space!.WELCOME 🤞")
      continue
    results=analysis(user_seq_input)
    print("\n" + "*"*45)
    print(f"01.Input DNA:{user_seq_input}")
    print(f"02.Complement strand:{results["template"]}")
    print(f"03.Messenger RNA:{results["mRNA"]}")
    print(f"04.Polypeptide chain:{results["polypeptide_chain"]}")
    print(f"05.GC percentage:{results["gc"]:.2f}%")
    print(f"06.Molecular weight of protein:{results["molw"]:.2f}g/mol")

    print(f"07.Melting temperature to sepatate duplex/circular gDNA:{results["tm"]:.2f}℃")
    print(f"08.Adaptive niche/Environment:{results["niche"]}")

```
