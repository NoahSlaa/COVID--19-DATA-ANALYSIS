```python
import pandas as np
import numpy as np
from sklearn.linear_model import LinearRegression
x=np.array([45,78,90,99,104,134,154,167,189]).reshape(-1, 1) # Corrected: np.array(...) and reshaped to 2D
y=np.array([67,90,97,105,112,145,167,189,199]) # Corrected: np.array(...)
model=LinearRegression()
model.fit(x,y)
print(f"The weight is:{model.coef_[0]:.2f}")
print(f"The intercept is:{model.intercept_:.2f}")
netx_hours=np.array([[789999999999999999]])
prediction=model.predict(netx_hours)
print(f"The next  hours will be: {prediction[0]:.2f}")
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
np.random.seed(455)
x=np.random.randint(1,100,size=(20,1))
y=(x*8)+ 20 + np.random.randint(10,50,size=(20,1))
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=455)
model=LinearRegression()
model.fit(x_train ,y_train)
y_predicted=model.predict(x_test)
mse=mean_squared_error(y_test,y_predicted)
r2=r2_score(y_test,y_predicted)
print(f"Mean squared error:{mse:.2f}")
print(f"R_SCORE:{r2:.2f}")
plt.scatter(x,y,color="blue",label="data points")
plt.plot(x,model.predict(x),color="red",label="Regression line")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error,root_mean_squared_error
np.random.seed(42)
noise=np.random.randint(-20,30,size=(100,))
x=np.random.randint(low=[50,1,1],high=[300,6,20],size=(100,3))
y=(x[:,0]*2)+(x[:,1]*3)+(x[:,2]*4)+noise
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
x_test,x_train,y_test,y_train=train_test_split(x,y,test_size=0.2,random_state=42)
model=LinearRegression()
model.fit(x_train,y_train)
y_predicted=model.predict(x_test)
print(f"The weight is:{model.coef_[0]:.2f}")
print(f"weight in 02:{model.coef_[1]:.2f}")
print(f"The intercept is:{model.intercept_:.2f}")
mse=mean_squared_error(y_test,y_predicted)
r2_score=r2_score(y_test,y_predicted)
mae=mean_absolute_error(y_test,y_predicted)
rmse=root_mean_squared_error(y_test,y_predicted)
print(f"Mean squared error:{mse:.2f}")
print(f"R_SCORE:{r2_score:.2f}")
print(f"Mean absolute error:{mae:.2f}")
print(f"Root mean squared error:{rmse:.2f}")
#
```
